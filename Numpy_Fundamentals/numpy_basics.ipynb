Vector operations:
While scalar operations involve an array and a single number, vector operations (also called element-wise operations) involve two arrays of the same shape.

In NumPy, when you perform an operation on two vectors, it applies the operation to the corresponding elements at each position. This is the heart of "vectorization"—it allows you to avoid writing for loops in Python, making your code much faster and cleaner.

If you have Vector A= $[a_1, a_2]  
        and Vector B $[b_1, b_2], 
        the operation happens positionally:
        A + B = [a_1 + b_1,  a_2 + b_2]

```python
arr_1= np.arange(1,17).reshape(4,4)
arr_2= np.arange(17,33).reshape(4,4)
```

axis()---->In NumPy and Pandas, the axis parameter is the "direction" in which an operation is performed.

The Rule of Thumb...

axis=0: Think Vertical (Down the rows)---Column wise operation

axis=1: Think Horizontal (Across the columns) ---Row wise operation

```python
np.max(array,axis=0) #axis=0 means we want to get the maximum value of each column. Here we are getting the maximum value of each column in the array.
```

```python
np.max(array,axis=1) #axis=1 means we want to get the maximum value of each row. Here we are getting the maximum value of each row in the array.
```
