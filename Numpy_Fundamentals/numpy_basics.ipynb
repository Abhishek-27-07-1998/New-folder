arange()----> provides the values over a certain range

```python
np.arange(1,15)
np.arange(5,10)
np.arange(2,21,2) #----> 3rd parameter is step size which means it will generate numbers from 2 to 20 with a step size of 2
```

```python
arange_data=np.arange(0,16)
arange_data.reshape(4,4)  #----> reshape function is used to change the shape of the array. Here we are changing the shape of the array from (16,) to (4,4) which means we are creating a 2D array with 4 rows and 4 columns.
arange_data.reshape(2,8) #----> reshape function is used to change the shape of the array. Here we are changing the shape of the array from (16,) to (2,8) which means we are creating a 2D array with 2 rows and 8 columns.
arange_data.reshape(8,2) #----> reshape function is used to change the shape of the array. Here we are changing the shape of the array from (16,) to (8,2) which means we are creating a 2D array with 8 rows and 2 columns.
arange_data.reshape(2,4,2) #----> reshape function is used to change the shape of the array. Here we are changing the shape of the array from (16,) to (2,4,2) which means we are creating a 3D array with 2 blocks, 4 rows and 2 columns.
```

#linspace()----> It make sure that every value within the given range going to have similar distance from each other

```python
np.linspace(-10,10,5) #linspace function is used to create an array of evenly spaced numbers between a specified range. Here we are creating an array of 5 evenly spaced numbers between -10 and 10. The 3rd parameter is the number of evenly spaced numbers we want to create.
np.arange(-10,10,5) #arange function is used to create an array of numbers between a specified range with a specified step size. Here we are creating an array of numbers between -10 and 10 with a step size of 5. The 3rd parameter is the step size which means it will generate numbers from -10 to 10 with a step size of 5.
```

#Shape attribute provides us with the idea about rows and columns in our dataset or in array

```python
array_array=np.array([[1,2,3],[4,5,6]]) #creating a 2D array with 2 rows and 3 columns.
array_array.shape #shape function is used to get the shape of the array. Here we are getting the shape of the array which is (2,3) which means we have 2 rows and 3 columns in our array.
```

Size()----> Provides total number of elements present inside our dataset or array

```python
array= np.array(range(5,25))
size=array.size #size function is used to get the number of elements in the array. Here we are getting the number of elements in the array which is 20.
size
```
