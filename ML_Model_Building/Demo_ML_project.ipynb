```python
st=st[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure',
        'CGPA', 'Study Satisfaction',
       'Sleep Duration', 'Dietary Habits',
        'Work/Study Hours',
       'Financial Stress', 'Family History of Mental Illness', 'Depression']]
```

#Label Encoding--Convert categorical data into numerical data

```python
from sklearn.preprocessing import LabelEncoder
l=LabelEncoder()
st['Gender']=l.fit_transform(st['Gender'])
```

```python
st['City']=l.fit_transform(st['City'])
```

```python
st['Profession']=l.fit_transform(st['Profession'])
```

```python
st['Dietary Habits']=l.fit_transform(st['Dietary Habits'])
```

```python
st['Family History of Mental Illness']=l.fit_transform(st['Family History of Mental Illness'])
```

```python
st['Sleep Duration']=l.fit_transform(st['Sleep Duration'])
```

```python
st.head()
```

```python
st['Financial Stress'].value_counts()
```

```python
st=st[st['Financial Stress']!='?']
```

```python
st['Financial Stress']=st['Financial Stress'].astype('Float32')
```

# Sepearating input and output data

```python
st.columns
```

```python
X=st[['Gender', 'Age', 'City', 'Profession', 'Academic Pressure', 'CGPA',
       'Study Satisfaction', 'Sleep Duration', 'Dietary Habits',
       'Work/Study Hours', 'Financial Stress',
       'Family History of Mental Illness']]
```

```python
Y=st['Depression']
```

# Feature Scaling


```python
from sklearn.preprocessing import StandardScaler
s=StandardScaler()
```

```python
#fit_transform() method in machine learning is used to calculate the necessary transformation parameters 
# (e.g., mean and standard deviation for scaling) from the training data and apply the transformation 
# to that data simultaneously
X=s.fit_transform(X)
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=42) 
#function is used to split the data into training and testing sets.
```

```python
X_train.shape
```

```python
X_test.shape
```

# Logistic regression

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
lr.fit(X_train,Y_train)
```

```python
y_predict=lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test,y_predict)
```

```python
from sklearn.tree import DecisionTreeClassifier
tree=DecisionTreeClassifier(max_depth=5)
```

```python
tree.fit(X_train,Y_train)
```

```python
y_predict=tree.predict(X_test)
```

```python
accuracy_score(Y_test,y_predict)
```
